## e) A las de tipo Object, recategorícelas a categorías numéricas.
Al final verifique lo realizado y que todas sus variables ahora son de tipo numéricas. Utilizando código, ¿Cuántas casos posee por cada categoría de cada variable recategorizada? Luego vuelque en su informe todo lo realizado y a lo que llegó.

```python
#e) A las de tipo Object, recategorícelas a categorías numéricas.
cols_object = data.select_dtypes(include=['object']).columns.tolist() #Contador
for col in cols_object:
    data[col] = data[col].astype('category').cat.codes #Recategorización

print(f"Recategorización de columnas de tipo Object a categorías numéricas:\n")
print(f"Número de columnas de tipo Object recategorizadas: {len(cols_object)}")
print(f"Número de columnas numéricas: {data.select_dtypes(include=['number']).shape[1]}")
```

```python
print(data.dtypes)
```

```python
# Verificar que todas las variables ahora son de tipo numéricas
all_numeric = data.select_dtypes(include=['number']).shape[1] == data.shape[1]
print(f"Todas las variables son ahora de tipo numéricas: {'Sí' if all_numeric else 'No'}")
```

```python
# ¿Cuántas casos posee por cada categoría de cada variable recategorizada?
category_counts = {}
for col in cols_object:
    category_counts[col] = data[col].value_counts().to_dict()

# Para trabajar las cantidades:
data_for_df = []

# Iterar sobre las columnas y sus conteos
for col, counts in category_counts.items():
    for category, count in counts.items():
        data_for_df.append([col, category, count])

#Se crea el Dataframe para contar los casos
df = pd.DataFrame(data_for_df, columns=['Variable', 'Categoría', 'Conteo'])

df_grouped_sum = df.groupby(['Variable', 'Categoría'])['Conteo'].sum().reset_index()

recate = f"Recategorización de columnas de tipo Object a categorías numéricas:\n"

for variable, group in df_grouped_sum.groupby(['Variable']):
    total_count = group['Conteo'].count()
    recate += f"De la variable '{variable}' tiene {total_count} casos contados.\n"

print(recate)
```

## l) Por cada columna que posee valores faltantes, cree una nueva nombrándola ‘MARCA_XX’
(en XX vaya poniendo nombres diferentes) en el DataFrame que valga 1 si posee valor faltante y sino que valga 0. Verifique en cada caso que la cantidad de valores faltantes que posee la variable original sea la misma que ahora vale uno en la variable creada.

```python
# Para l) a través de un iterador y un condicional creamos la columna nueva y asignamos si corresponde 0 si existe valor o 1 si no existe
for col in new_df.columns:
    if new_df[col].isnull().sum() > 0:
        nueva_columna = 'MARCA_' + col
        new_df[nueva_columna] = new_df[col].isnull().astype(int)

# Conteo por columnas marcadas como 0 y 1
for col in new_df.columns:
    if col.startswith('MARCA_'):
        original_col = col.replace('MARCA_', '')
        print(f"Valores faltantes en {original_col}: {new_df[col].sum()}")
        print(f"Valores faltantes en MARCA_{original_col}: {new_df[original_col].isnull().sum()}")
```
